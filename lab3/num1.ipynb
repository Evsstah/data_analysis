```python
import numpy as np

# 1. Загрузка данных
# данные потребления
consumption_numeric = np.genfromtxt(
    'global-electricity-consumption.csv',
    delimiter=',',
    skip_header=1,
    usecols=range(1, 31),
    missing_values='--',
    filling_values=np.nan
)

# названия стран
countries = np.genfromtxt(
    'global-electricity-consumption.csv',
    delimiter=',',
    skip_header=1,
    usecols=[0],
    dtype=str,
    encoding='utf-8'
)

# данные производства
generation_numeric = np.genfromtxt(
    'global-electricity-generation.csv',
    delimiter=',',
    skip_header=1,
    usecols=range(1, 31),
    missing_values='--',
    filling_values=np.nan
)

years = np.arange(1992, 2022)


# 2. Среднее производство и потребление для каждой страны за последние 5 лет
last_5_years_indices = [25, 26, 27, 28, 29]  # Индексы 2017-2021 годов

# Среднее по годам для каждой страны
avg_generation_per_country = np.nanmean(generation_numeric[:, last_5_years_indices], axis=1)
avg_consumption_per_country = np.nanmean(consumption_numeric[:, last_5_years_indices], axis=1)

# 3.1 Суммарное потребление по всем странам за каждый год
total_consumption_per_year = np.nansum(consumption_numeric, axis=0)

# 3.2 Максимальное производство одной страной за один год
max_generation_single = np.nanmax(generation_numeric)

# 3.3 Страны, производящие >500 млрд кВт*ч в среднем за последние 5 лет
high_generation_countries = countries[avg_generation_per_country > 500]

# 3.4 10% стран с наибольшим потреблением за последние 5 лет
threshold_consumption = np.nanquantile(avg_consumption_per_country, 0.9)
top_10_percent_consumption = countries[avg_consumption_per_country >= threshold_consumption]

# 3.5 Страны, увеличившие производство в 2021 году более чем в 10 раз по сравнению с 1992
generation_1992 = generation_numeric[:, 0]  # 1992 год
generation_2021 = generation_numeric[:, -1]  # 2021 год
increase_ratio = generation_2021 / generation_1992

valid_indices = ~np.isnan(increase_ratio) & ~np.isinf(increase_ratio)
countries_10x_increase = countries[valid_indices & (increase_ratio > 10)]

# 3.6 Страны, потратившие в сумме >100 млрд кВт*ч и потреблявшие больше, чем производили
total_consumption_all_years = np.nansum(consumption_numeric, axis=1)
total_generation_all_years = np.nansum(generation_numeric, axis=1)
deficit_countries = countries[
    (total_consumption_all_years > 100)
    (total_consumption_all_years > total_generation_all_years)
]

# 3.7 Страна с наибольшим потреблением в 2020 году
consumption_2020 = consumption_numeric[:, -2]
max_consumption_2020_index = np.nanargmax(consumption_2020)
country_max_consumption_2020 = countries[max_consumption_2020_index]


print("РЕЗУЛЬТАТЫ АНАЛИЗА ЭЛЕКТРОЭНЕРГЕТИКИ")

print("\n3.1 Суммарное потребление по годам (млрд кВт*ч):")
for year, consumption in zip(years, total_consumption_per_year):
    print(f"{year}: {consumption:.1f}")

print(f"\n3.2 Максимальное производство одной страной за год: {max_generation_single:.2f} млрд кВт*ч")

print(f"\n3.3 Страны с производством >500 млрд кВт*ч в среднем за последние 5 лет:")
for country in high_generation_countries:
    idx = np.where(countries == country)[0][0]
    avg_val = avg_generation_per_country[idx]
    print(f"  {country}: {avg_val:.1f} млрд кВт*ч")

print(f"\n3.4 10% стран с наибольшим потреблением за последние 5 лет:")
for country in top_10_percent_consumption:
    idx = np.where(countries == country)[0][0]
    avg_val = avg_consumption_per_country[idx]
    print(f"  {country}: {avg_val:.1f} млрд кВт*ч")

print(f"\n3.5 Страны, увеличившие производство >10 раз с 1992 по 2021:")
for country in countries_10x_increase:
    idx = np.where(countries == country)[0][0]
    ratio = increase_ratio[idx]
    gen_1992_val = generation_1992[idx]
    gen_2021_val = generation_2021[idx]
    print(f"  {country}: {gen_1992_val:.1f} → {gen_2021_val:.1f} (в {ratio:.1f} раз)")

print(f"\n3.6 Страны с суммарным потреблением >100 млрд кВт*ч и дефицитом производства:")
for country in deficit_countries:
    idx = np.where(countries == country)[0][0]
    total_cons = total_consumption_all_years[idx]
    total_gen = total_generation_all_years[idx]
    deficit = total_cons - total_gen
    print(f"  {country}: потребление {total_cons:.1f}, производство {total_gen:.1f} (дефицит: {deficit:.1f})")

print(f"\n3.7 Страна с наибольшим потреблением в 2020 году:")
max_consumption_value = consumption_2020[max_consumption_2020_index]
print(f"  {country_max_consumption_2020}: {max_consumption_value:.1f} млрд кВт*ч")
```
